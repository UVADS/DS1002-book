# Control Structures

R code needs looping and conditionals just like Python does. Here are some examples and practice exercises for you:

## Conditions

R uses a fairly similar model to Python. The `if` and `else if` controls require a logical test that results in a `TRUE`|`FALSE` and the `else` remains a catch-all that has no condition.

In terms of syntax, the main differences with Python are: 

- Conditions are included between parentheses.
- There's no `:` after the condition statement to indicate the code block to run.
- Code block to be run is enclosed between `{}`, instead of by indentation.
- `elif` is called `else if` in R.

On the other hand, comparison operators in R are the following:

| Operator  | Name  | Example  |
|---|---|---|
| `==` | Equals  | `x == y` |
| `!=` | Not equals  | `y != x` |
| `>`  | Greater than  | `x > y` |
| `<`  | Less than  | `y < x` |
| `>=` | Greater than or equal to  | `x >= y` |
| `<=` | Less than or equal to  | `y <= x` |



````r
x <- 14
y <- 120

if (y > x) {
  print("y is greater than x")
}
````

````r
x <- 33
y <- 33

if (y > x) {
  print("y is greater than x")
} else if (x == y) {
  print("x and y are equal")
}
````

````r
x <- 120
y <- 44

if (y > x) {
  print("y is greater than x")
} else if (x == y) {
  print("x and y are equal")
} else {
  print("x is greater than y")
}
````

````r
# input vector
x <- c(12, 9, 23, 14, 20, 1, 5)

# ifelse() function to determine odd/even numbers
ifelse(x %% 2 == 0, "EVEN", "ODD")
````

## Loops


### `while` loops


The syntax for `while` loops in R is the following:

```
while ( CONDITION ) {
STATEMENT1
STATEMENT2
ETC
}
```



````r
x <- 0
while ( x < 100 ) {
x <- x + 17
print( x )
}
print("I exited the loop, yeah!")
````

### `break`  - exit the loop

As in Python, it stops the loop entirely.

````r
# BREAK

i <- 1
while (i < 9) {
  print(i)
  i <- i + 1
  if (i == 5) {
    print("I'm all done!")
    break
  }
}
````

### `next` - stop the current iteration

It skips an interation of the loop and moves on to the next step in logic. In Python this corresponds to the `continue` keyword.

````r
# NEXT
i <- 0
while (i < 8) {
  i <- i + 1
  if (i == 5) {
    print("Here is where I pass")
    next
  }
  print(i)
}
````

### `for` loops

The syntax for `for` loops in R is the following:

```
for ( VAR in iterable ) {
STATEMENT1
STATEMENT2
ETC
}
```


In R, an iterable might be a numeric count or range (i.e. using `:` or `seq`), or through a series of elements in a list or vector, or for a numeric range/limit against a vector. 

````r
# loops 10x, counting 1-10.
for(i in 1:10) {
  print(i)
}
````

````r
# loops once for each item in a vector
x <- c("a", "b", "x", "y", "z")
for(i in x) {
  print(i)
}
````

````r
# loops in a range of items (3x) mapped against a vector.
# therefore loops 3x total against a, b, and x.

x <- c("a", "b", "x", "y", "z")

for(i in 1:3) {
  print(x[i])
}
````

## `seq_along()` 

This function has the same spirit of Python's `enumerate`: it generates an integer sequence based on the length of an object.

````r
x <- c("a", "b", "x", "y", "z")

for(ii in seq_along(x)) {
  print(ii)     # the integer of the vector item index
  print(x[ii])  # the actual value of the vector item
}
````

````r
# compare to this for-loop that skips integer/indexing and simply takes each value of the vector in sequence
for(letter in x) {
  print(letter)
}
````

````r
# What about iterating across a range, but only at certain intervals?
# Use a sequence from 1-30 but count in intervals of 3

for(i in seq(1, 30, 3)) {
  print(i)
}
````
